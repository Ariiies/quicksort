**Import Quicksort and defines an unsorted list.**

```python
from quicksort import quicksort 

data = [4,5,9,10,16,1,8,7,2,6,14]
```

**Show the data. This function uses sublists for the sort, what is means that no modifies the original list.**

```python
print("---Unsorted list---")
print(data)
print("---Sorted list in ascending sorting---")
print(quicksort(data))
print("---Sorted list in descending sorting---")
print(quicksort(data, True))
print("--- Original list ---")
print(data)
```
